Ваша задача - использовать вычисления Numpy для анализа этих данных и доказать, что ваши СrunchieMunchies - самый здоровый выбор для потребителей.

```python
calorie_stats = np.loadtxt("./data/cereal.csv", delimiter=",")
calorie_stats
```

```python
average_calories = calorie_stats.mean()
print(f"Mean: {average_calories}")
print(f"The average number of calories among competitors is {average_calories - 60} higher")
```

```python
calorie_stats_sorted = np.sort(calorie_stats)
print(calorie_stats_sorted)
```

```python
higher_avg = calorie_stats[calorie_stats > average_calories]
print(len(calorie_stats))
print(len(higher_avg))
```

```python
median_index = len(calorie_stats) // 2 + 1
median_calories = calorie_stats_sorted[median_index]
print("Median:", median_calories)
```

```python
percentiles = np.arange(0, 101, 5)
values = np.percentile(calorie_stats_sorted, percentiles)

nth_percentile = None

for p, val in zip(percentiles, values):
    print(f"{p} percentile: {val} calories")
    if val > 60 and nth_percentile is None:
        nth_percentile = p
```

```python
print("The smallest percentile the value of which is greater than 60: ",nth_percentile)
```

```python
more_calories = len(calorie_stats[calorie_stats > 60]) / len(calorie_stats) * 100
print(f"{more_calories} %")
```

```python
calorie_std = calorie_stats_sorted.std()
print("Std: ",calorie_std)
```

The CrunchieMunchies bar has 60 calories, which is much lower than most other bars. The average calorie content of competitors’ bars is about 107, and the median is 110. Around 96% of the bars have more calories than 60, so the CrunchieMunchies bar is a lighter option. Mycrunch could use this to market the bar as a lower-calorie snack
