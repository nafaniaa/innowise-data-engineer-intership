Задача 1: Подсчитать произведение ненулевых элементов на диагонали прямоугольной матрицы.
Например, для X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) ответ 3.

```python
X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]])
diag = np.diagonal(X)
non_zero = diag[diag != 0]
res = np.prod(non_zero)
print(res)

X_list = X.tolist()
res_list = 1
for i in range(min(len(X_list),len(X_list[0]))):
    if X_list[i][i] != 0:
        res_list *= X_list[i][i] 

print(res_list)
```

Задача 4: Реализовать кодирование длин серий (Run-length encoding). Для некоторого вектора x необходимо вернуть кортеж из двух векторов одинаковой длины. Первый содержит числа, а второй - сколько раз их нужно повторить.
Например, для x = np.array([2, 2, 2, 3, 3, 3, 5]) ответ (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
x = np.array([2, 2, 2, 3, 3, 3, 5])
print(np.unique(x,return_counts=True))

x_list = x.tolist()
count_dict = {}

for el in x_list:
    if el in count_dict.keys():
        count_dict[el] += 1
    else:
        count_dict[el] = 1

print(count_dict.keys(), count_dict.values())
```

Задача 5: Даны две выборки объектов - X и Y. Вычислить матрицу евклидовых расстояний между объектами. Сравните с функцией scipy.spatial.distance.cdist по скорости работы.

```python
import time
from scipy.spatial.distance import cdist
import math 

def euclidean_distances_numpy(X, Y):
    diff = X[:, np.newaxis, :] - Y[np.newaxis, :, :]
    return np.sqrt(np.sum(diff**2, axis=2))


X = np.random.rand(1000, 50)
Y = np.random.rand(1000, 50)


start = time.time()
D1 = euclidean_distances_numpy(X, Y)
t_numpy = time.time() - start


start = time.time()
D2 = cdist(X, Y)
t_scipy = time.time() - start

print("NumPy:", t_numpy)
print("SciPy:", t_scipy)


def euclidean_distances(X, Y):
    D = [] 
    for x in X: 
        row = []  
        for y in Y:  
            dist = math.sqrt(sum((xi - yi)**2 for xi, yi in zip(x, y)))
            row.append(dist)
        D.append(row)
    return D


X_list = [[1, 2], [3, 4]]
Y_list = [[5, 6], [7, 8], [9, 10]]

X_numpy = np.array(X_list)
Y_numpy = np.array(Y_list)

D = euclidean_distances(X_list, Y_list)
for row in D:
    print(row)

print(euclidean_distances_numpy(X_numpy, Y_numpy))
```
